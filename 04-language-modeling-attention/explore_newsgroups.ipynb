```python
from sklearn.datasets import fetch_20newsgroups
```

```python
CATEGORIES = ["rec.sport.hockey", "sci.space"]
```

```python
data = fetch_20newsgroups(subset="train")
for i, name in enumerate(data.target_names):
    print(i, name)
```

```python
import numpy as np
print("# number of posts:", len(data.data))                        
print("# posts per category: ", np.bincount(data.target))              
print("# first post, first 500 chars:\n", data.data[0][:500])                    
print("# its category: ", data.target_names[data.target[0]])     
```

```python
import re

def tokenize(text):
    return re.findall(r"[a-z']+", text.lower())[:100]
```

```python
train = tokenize(data.data[0][:500])

print(train)
```

```python
train_raw = fetch_20newsgroups(subset="train", categories=CATEGORIES, remove=("headers", "footers", "quotes"))
```
